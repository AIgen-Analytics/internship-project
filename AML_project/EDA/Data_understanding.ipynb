# Outlier Analysis

The transaction_amount feature was analyzed for potential outliers using the Interquartile Range (IQR) method.

```python
Q1 = transactions['transaction_amount'].quantile(0.25)
Q3 = transactions['transaction_amount'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = transactions[
    (transactions['transaction_amount'] < lower_bound) |
    (transactions['transaction_amount'] > upper_bound)
]

print("Number of Outliers:", len(outliers))
```

        ▼

## Observations

- Customer dataset acts as the central dataset.
- Accounts are linked to customers using customer_cif.
- Devices are linked to customers using customer_cif.
- Transactions are linked to accounts using account numbers.
- These relationships help trace customer activities and transaction flows for AML analysis.

```python
# ── Cell 1: Fraud vs Normal — transaction amount comparison ──────────────
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Log-scale distribution
for label, color in zip([0, 1], ['steelblue', 'crimson']):
    subset = transactions[transactions['is_aml'] == label]['transaction_amount']
    axes[0].hist(np.log1p(subset), bins=60, alpha=0.6,
                 label='Normal' if label==0 else 'AML', color=color)
axes[0].set_title('Transaction Amount Distribution (log scale)')
axes[0].set_xlabel('log(1 + amount)')
axes[0].legend()

# Box plot
transactions['is_aml_label'] = transactions['is_aml'].map({0: 'Normal', 1: 'AML'})
transactions.boxplot(column='transaction_amount', by='is_aml_label', ax=axes[1])
axes[1].set_title('Amount by AML Status')
axes[1].set_yscale('log')
plt.tight_layout()
plt.show()
```

```python
# ── Cell 2: AML Typology distribution ────────────────────────────────────
aml_only = transactions[transactions['is_aml'] == 1]
typology_counts = aml_only['aml_typology'].value_counts()

plt.figure(figsize=(12, 5))
typology_counts.plot(kind='bar', color='crimson', edgecolor='black')
plt.title('AML Typology Distribution (72,854 fraud cases)')
plt.xlabel('Typology')
plt.ylabel('Count')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()

print(typology_counts)
print(f"\nFraud rate: {transactions['is_aml'].mean()*100:.1f}%")
```

```python
# ── Cell 3: FIS Band and Alert Level breakdown ────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

transactions['fis_band'].value_counts().plot(kind='bar', ax=axes[0], color='darkorange')
axes[0].set_title('Fraud Intensity Score Band')
axes[0].tick_params(axis='x', rotation=30)

transactions['alert_level'].value_counts().plot(kind='bar', ax=axes[1], color='steelblue')
axes[1].set_title('Alert Level Distribution')
axes[1].tick_params(axis='x', rotation=30)

plt.tight_layout()
plt.show()
```

```python
# ── Cell 4: Top features correlated with is_aml ──────────────────────────
num_cols = transactions.select_dtypes(include='number').columns.tolist()
corr_with_target = transactions[num_cols].corr()['is_aml'].drop('is_aml').abs().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
corr_with_target.head(20).plot(kind='barh', color='purple')
plt.title('Top 20 features correlated with is_aml')
plt.xlabel('Absolute Correlation')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print(corr_with_target.head(20))
```
